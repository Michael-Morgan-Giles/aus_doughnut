## Food

### Prevalence of undernourishment
The prevalence of undernourishment (PoU) is an estimate of the proportion of the population whose habitual food consumption is insufficient to provide the dietary energy levels that are required to maintain a normal active and healthy life. It is expressed as a percentage. This indicator will measure progress towards SDG Target 2.1.

https://www.fao.org/sustainable-development-goals-data-portal/data/indicators/2.1.1-prevalence-of-undernourishment/en

This measure is not very interesting for Australia because the actual value is no doubt less than the 2.5 estimate provided. But I assume the precision of the estimate was not high enough to provide a lower estimate with enough confidence. I will not include it in the analysis.

```python
under_nourish = pd.read_csv(input_path + "social/1_food/prevelance_undernourishment.csv")
under_nourish = under_nourish[['TIME_PERIOD','OBS_VALUE']]

under_nourish['Date'] = under_nourish['TIME_PERIOD']
under_nourish['domain'] = "social"
under_nourish['dimension'] = "food"
under_nourish['indicator'] = "prev_undernourishment"
under_nourish['value'] = under_nourish['OBS_VALUE'].str.replace("<","").astype(float)

under_nourish.dropna()
```

### Population with moderate to severe food insecurity

SDG indicator 2.1.2 Prevalence of moderate or severe food insecurity in the population, based on the Food Insecurity Experience Scale

https://www.fao.org/sustainable-development-goals-data-portal/data/indicators/212-prevalence-of-moderate-or-severe-food-insecurity-in-the-population-based-on-the-food-insecurity-experience-scale/en

It is a quantitative measure of food insecurity derived from 8 qualitative and quantitative questions regarding sufficient access to food. It uses the self-reported Food Insecurity Experience Scale (FIES) standard.

Because it is self-reported it can be a slightly biased measure due to coginitive heuristics. However, it conveys what we are trying to measure.

The social foundation for food insecurity is zero.

```python
food_insecure = pd.read_csv(input_path + "social/1_food/food_insecurity.csv")

food_insecure = food_insecure[['TIME_PERIOD','OBS_VALUE']]

food_insecure['Date'] = food_insecure['TIME_PERIOD']
food_insecure['domain'] = "social"
food_insecure['dimension'] = "food"
food_insecure['indicator'] = "food_insecurity"
food_insecure['value'] = food_insecure['OBS_VALUE']
food_insecure['boundary'] = 0

food_insecure[final_col_list].dropna().sort_values("Date").reset_index(drop=True)
```

```python
food_insecure[final_col_list].dropna().sort_values("Date").reset_index(drop=True).to_csv(input_path +  "/social/clean/food_insecure_clean.csv", index = False)
```

Source: https://unstats.un.org/sdgs/dataportal/database

The World Health Organisation has created the Universal Health Coverage (UHC) index. UHC is a geometric mean of 14 different proxy variables. The boundary set in the proportion of the worlds population living in a country with a UHC less than 60.

### Social Housing Wait List

Therefore, I choose to adopt a measure of how many people in the greatest need are on waiting lists for social housing. 

According to the Australian Institute of Health and Wellbeing; *"In all states and territories, access to social housing is managed through the use of waiting lists with priority given to those considered to be high priority applicants".*

Therefore, I have defined a measure of the percentage of people on the waiting list in the "greatest need" *(defined slightly differently across each state/territory)*, as a proportion of all people on the waitlist. 

Source: https://www.aihw.gov.au/reports/housing-assistance/housing-assistance-in-australia-2023/contents/households-and-waiting-lists

### No Difference in Imprisonment Rates between Indigenous and Non-Indigenous Australians

There was no measure included for racial equality in the doughnut study. In Australia there is a unconscionably high gap in imprisonment rates between Indigenous and Non-indigenous Australians. 

Source: https://www.pc.gov.au/closing-the-gap-data/dashboard/outcome-area/criminal-justice/

Eventually we push for no prisons, however more immediately, the gap between indigenous and non-indigenous people in prisons per 100,000 people should be zero by 2030. This seems like the most obvious racial equality index to measure for a doughnut in Australia.

I calculate the percentage difference between the imprissonment rate per 100,000 people using the productivity comissions data as such:

$$ \Delta_{\text{Imprisonment Rate}}\% = \frac{\text{Indigenous Prisoners per 100000}_t - \text{Non-indigenous Prisoners per 100000}_t}{\text{Non-indigenous Prisoners per 100000}_t}$$

That is for their to be equality in imprisonment rates, $ \Delta_{\text{Imprisonment Rate}}\% = 0$. Therefore, the social foundation is set at zero. To create a ratio, bounded by 0 and 1, a worst case scenario will be adopted. This is further described in the analysis script. 

This is obviously an insufficient measure for true racial equality. More needs to be added to the social foundation to improve this aspect of the doughnut.

```python
imprisonment = pd.read_csv(input_path + "social/9_equality/imprisonment_rates.csv")
imprisonment = imprisonment[(imprisonment['Description1'] == 'Age-standardised imprisonment rate per 100000 adults at 30 June') &
                           (imprisonment['Description7'] == "Actual")][['Year', 'Indigenous_Status', 'Aust']]

imprisonment['Aust'] = imprisonment['Aust'].astype(float)
wide_imprison = pd.pivot_table(imprisonment, 
                               columns = ['Indigenous_Status'], 
                               index = ['Year'], 
                               values = "Aust").reset_index()
wide_imprison
```

```python
imprisonment = pd.read_csv(input_path + "social/9_equality/imprisonment_rates.csv")
imprisonment = imprisonment[(imprisonment['Description1'] == 'Age-standardised imprisonment rate per 100000 adults at 30 June') &
                           (imprisonment['Description7'] == "Actual")][['Year', 'Indigenous_Status', 'Aust']]

imprisonment['Aust'] = imprisonment['Aust'].astype(float)
wide_imprison = pd.pivot_table(imprisonment, 
                               columns = ['Indigenous_Status'], 
                               index = ['Year'], 
                               values = "Aust").reset_index()

wide_imprison['imprison_gap_per_100000'] = wide_imprison['Aboriginal and Torres Strait Islander people'] - wide_imprison['Non-Indigenous people']
wide_imprison['perc_imprison_gap_per_100000'] = wide_imprison['imprison_gap_per_100000']/wide_imprison['Non-Indigenous people']
wide_imprison['Date'] = wide_imprison['Year']
wide_imprison['domain'] = "social"
wide_imprison['dimension'] = "equality"
wide_imprison['indicator'] = "perc_more_ind_prison_compared_to_non_ind_prisons_per_100000"
wide_imprison['value'] = wide_imprison['perc_imprison_gap_per_100000'].astype(float)
wide_imprison['boundary'] = 0
wide_imprison['ratio'] = (wide_imprison['value'] - 1)/wide_imprison['value']

wide_imprison[final_col_list + ['ratio']]
```
